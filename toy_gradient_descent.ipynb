```python
import numpy as np
import matplotlib.pyplot as plt
 
np.random.seed(0)
x = np.random.rand(100)
y_true = 2 * x + 1 + np.random.randn(100) * 0.1
 
w, b = 0.0, 0.0
lr = 0.1
losses = []
for step in range(200):
    y_pred = w * x + b
    loss = np.mean((y_pred - y_true) ** 2)
    dw = np.mean(2 * (y_pred - y_true) * x)
    db = np.mean(2 * (y_pred - y_true))
    w -= lr * dw
    b -= lr * db
    losses.append(loss)
 
print(w, b)
plt.plot(losses)
plt.show()
```
